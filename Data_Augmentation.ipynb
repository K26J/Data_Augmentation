```python
from tensorflow import keras
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img= image.load_img('/content/test_cat.jpg', target_size= (256,256))
```

```python
img
```

```python
datagen= ImageDataGenerator(
    rotation_range= 40, # rotation_angle of image upto 40 degree
    shear_range= 0.2,
    zoom_range= 0.2,
    horizontal_flip= True,
    width_shift_range= 0.2,
    height_shift_range= 0.2,
    fill_mode= 'nearest'
)
```

```python
img= image.img_to_array(img)# Converting image to numpy array
```

```python
type(img)
```

```python
img.shape
```

```python
input_batch= img.reshape(1,256,256,3)
```

```python
import os

i= 0

os.makedirs('aug', exist_ok=True)
for output in datagen.flow(input_batch, batch_size= 1, save_to_dir= 'aug'):

  i= i+1
  if i==20:
    break
```
